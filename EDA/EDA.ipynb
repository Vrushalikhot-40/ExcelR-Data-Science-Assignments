## EXPLORATORY DATA ANALYSIS ON A DATASET

#### 1.	Data Cleaning and Preparation

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("Cardiotocographic.csv")
```

```python
df.info()
```

```python
df.shape
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
df.fillna(df.median(numeric_only=True), inplace=True)
```

```python
df.dtypes
```

```python
sns.boxplot(data=df)
plt.xticks(rotation=90)
```

#### 2.	Statistical Summary:

```python
df.mean()
```

```python
df.median()
```

```python
df.std()
```

```python
Q1 = df.quantile(0.25)
Q3 = df.quantile(0.75)
IQR = Q3 - Q1
```

##### Observation

##### The mean and median values help identify skewness. Standard deviation and IQR provide information about variability and spread.

#### 3.	Data Visualization:

```python
plt.figure(figsize=(15,8))
sns.boxplot(data=df)
plt.xticks(rotation=90)
plt.title("Boxplot for Outlier Detection")
plt.show()
```

```python
df.hist(figsize=(15,12))
plt.suptitle("Histogram of Numerical Variables")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12,8))
sns.heatmap(df.corr(), annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```

```python
selected_columns = ['LB','AC','FM','UC','ASTV','MSTV']

sns.pairplot(df[selected_columns])
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.violinplot(y=df['LB'])
plt.title("Violin Plot of Baseline Fetal Heart Rate (LB)")
plt.show()
```

#### 4.	Pattern Recognition and Insights:

```python
plt.figure(figsize=(8,6))
sns.scatterplot(x='AC', y='FM', data=df)
plt.title("Accelerations vs Fetal Movements")
plt.show()
```
