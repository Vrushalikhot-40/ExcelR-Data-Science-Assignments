## 1. Hypothesis Testing

```python
from scipy.stats import norm
import math

# Given data
sample_mean = 3050
x = 600
n = 25

# Theoretical mean cost
mu = 1000 + 5 * x

# Population standard deviation
sigma = 5 * 25

# Calculate Z statistic
z = (sample_mean - mu) / (sigma / math.sqrt(n))

# Critical value for alpha = 0.05 (right-tailed test)
critical_value = norm.ppf(0.95)

print("Theoretical Mean =", mu)
print("Z Statistic =", round(z, 4))
print("Critical Value =", round(critical_value, 4))

if z > critical_value:
    print("Reject H0")
    print("There is evidence that weekly costs are higher.")
else:
    print("Fail to Reject H0")
    print("Not enough evidence that costs are higher.")
```

## 2. Confidence Interval

#### a. Build 99% Confidence Interval Using Sample Standard Deviation

```python
import numpy as np
from scipy.stats import t

data = [1.13, 1.55, 1.43, 0.92, 1.25,
        1.36, 1.32, 0.85, 1.07, 1.48,
        1.20, 1.33, 1.18, 1.22, 1.29]

n = len(data)
mean = np.mean(data)
s = np.std(data, ddof=1)

# 99% confidence level
t_value = t.ppf(0.995, n-1)

margin = t_value * (s / np.sqrt(n))

lower = mean - margin
upper = mean + margin

print("Sample Mean =", round(mean, 4))
print("99% Confidence Interval:")
print("(", round(lower, 4), ",", round(upper, 4), ")")
```

#### b. Build 99% Confidence Interval Using Known Population Standard Deviation

```python
import numpy as np
from scipy.stats import norm

data = [1.13, 1.55, 1.43, 0.92, 1.25,
        1.36, 1.32, 0.85, 1.07, 1.48,
        1.20, 1.33, 1.18, 1.22, 1.29]

n = len(data)
mean = np.mean(data)

sigma = 0.2

z = norm.ppf(0.995)

margin = z * (sigma / np.sqrt(n))

lower = mean - margin
upper = mean + margin

print("Sample Mean =", round(mean, 4))
print("99% Confidence Interval:")
print("(", round(lower, 4), ",", round(upper, 4), ")")
```
