#### 1.Create a list of 5 integers. Perform and print the result of the following operations: append, extend, insert, remove, pop, clear, index, count, sort, and reverse.

```python
my_list = [2,3,4,5,6]
print(my_list)

my_list.append(8)
print("After append:",my_list)

my_list.extend([10,20])
print("After extend:",my_list)

my_list.insert(3, 9)
print("After insert:",my_list)

my_list.remove(5)
print("After remove:",my_list)

pop_value = my_list.pop(3)
print("After pop:",pop_value)

index_value = my_list.index(6)
print("After index:",index_value)

count_value = my_list.count(2)
print("After count:",count_value)

my_list.sort()
print("After sort:",my_list)

my_list.reverse()
print("After reverse:",my_list)

my_list.clear()
print("After clear:",my_list)
```

## Section C: Exploring Pandas

#### 11.Create a small DataFrame manually with 10 rows and columns: 'Name', 'Age', 'City', and 'Salary'. Then:
####  - Use .info(), .describe()
####  - Select 'Name' and 'City' columns
####  - Drop 'City' column
####  - Fill any missing values in 'Salary' column with the mean
####  - Remove any duplicate rows


```python
import numpy as np
import pandas as pd

data = {
    'Name': ['Riya', 'Rasika', 'Shreya', 'Saloni', 'Payal', 'Shruti', 'Rutuja', 'Arati', 'Anamika', 'Samrudhi'],
    'Age': [26, 22, 23, 23, 23, 24, 24, 24, 24, 24,],
    'City': ['Mumbai', 'Pune', 'Delhi', 'Chennai', 'Pune', 'Mumbai', 'Bengaluru', 'Delhi', 'Chennai', 'Pune'],
    'Salary': [np.nan, 50000, 45000, 35000, 42000, 21000, 55000, 34000, np.nan, 52000]
}

df = pd.DataFrame(data)

df.info()
df.describe()

print(df[['Name', 'City']])
df = df.drop('City', axis=1)

df['Salary'] = df['Salary'].fillna(df['Salary'].mean())

df = df.drop_duplicates()

print(df)
```
